```python
import seaborn as sns
%matplotlib inline
iris=sns.load_dataset('iris')
iris
```

```python
iris['species'].unique()
```

```python
sns.pairplot(iris)
```

```python

import matplotlib.pyplot as plt
g=sns.PairGrid(iris)
g.map_diag(sns.distplot)
g.map_upper(plt.scatter)
g.map_lower(sns.kdeplot)
```
